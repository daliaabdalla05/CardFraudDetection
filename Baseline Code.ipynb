```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("mlg-ulb/creditcardfraud")

print("Path to dataset files:", path)
```

```python
import pandas as pd
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import precision_recall_curve, average_precision_score

df = pd.read_csv("/kaggle/input/creditcardfraud/creditcard.csv").sort_values("Time")
split = int(len(df) * 0.75)
train, test = df.iloc[:split], df.iloc[split:]

X_train, y_train = train.drop(columns="Class"), train["Class"]
X_test, y_test = test.drop(columns="Class"), test["Class"]

scaler = StandardScaler().fit(X_train[["Amount", "Time"]])
for X in (X_train, X_test):
    X[["Amount", "Time"]] = scaler.transform(X[["Amount", "Time"]])

clf = LogisticRegression(max_iter=1000, class_weight="balanced").fit(X_train, y_train)
scores = clf.predict_proba(X_test)[:, 1]
print("PR-AUC:", average_precision_score(y_test, scores))
precision, recall, thr = precision_recall_curve(y_test, scores)
```
